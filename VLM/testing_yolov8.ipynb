```python
import cv2
from ultralytics import YOLO

# 1. Load the YOLOv8 model (n = nano, the fastest version)
model = YOLO('yolov8n.pt') 

# 2. Initialize the webcam
# '0' is usually the default built-in camera
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Error: Could not open webcam.")
    exit()

print("Press 'Space' to take a picture and run inference, or 'q' to quit.")

while True:
    # Capture frame-by-frame for the live preview
    ret, frame = cap.read()
    if not ret:
        break

    # Display the live preview
    cv2.imshow('Webcam Live Preview', frame)

    key = cv2.waitKey(1) & 0xFF
    
    # Press 'Space' to "Take a Picture"
    if key == ord(' '):
        print("Picture taken! Running inference...")
        
        # 3. Run inference on the captured frame
        # results is a list of Result objects (one per image)
        results = model(frame)

        # 4. Show the results
        # This will open a new window with the detections plotted
        for r in results:
            # Displays the image with bounding boxes
            r.show() 
            
            # Optional: Save the result to a file
            # r.save(filename='result.jpg')
            
        print("Inference complete.")

    # Press 'q' to quit the application
    elif key == ord('q'):
        break

# Clean up
cap.release()
cv2.destroyAllWindows()
```
